# **Clean Missing Value**

```python

# Make a copy of the original dataset
df_clean = df.copy()


```

```python
# Convert blank or whitespace-only values into missing values
import numpy as np
df_clean = df_clean.replace(r'^\s*$', np.nan, regex=True)
```

```python
# Check missing values again
print(df_clean.isnull().sum())
```

```python

# Fill numerical columns with their median
df_clean['Delivery_person_Age'] = df_clean['Delivery_person_Age'].fillna(
    df_clean['Delivery_person_Age'].median()
)

df_clean['Delivery_person_Ratings'] = df_clean['Delivery_person_Ratings'].fillna(
    df_clean['Delivery_person_Ratings'].median()
)

df_clean['multiple_deliveries'] = df_clean['multiple_deliveries'].fillna(
    df_clean['multiple_deliveries'].median()
)

# Fill categorical columns with their most frequent value
categorical_columns = [
    'Weather_conditions',
    'Road_traffic_density',
    'Festival',
    'City'
]

for col in categorical_columns:
    df_clean[col] = df_clean[col].fillna(df_clean[col].mode()[0])

# Keep missing order times for now; investigate them before deciding how to handle them

# Check remaining missing values
print(df_clean.isnull().sum())
```

# **Analyze delivery time**

```python

# Basic delivery time analysis

print("Average delivery time:",
      round(df_clean['Time_taken (min)'].mean(), 2), "minutes")

print("Minimum delivery time:",
      df_clean['Time_taken (min)'].min(), "minutes")

print("Maximum delivery time:",
      df_clean['Time_taken (min)'].max(), "minutes")

print("Median delivery time:",
      df_clean['Time_taken (min)'].median(), "minutes")
```

# **Analyze delivery time by traffic**

```python

# Average delivery time by road traffic density

traffic_analysis = (
    df_clean.groupby('Road_traffic_density')['Time_taken (min)']
    .agg(['mean', 'count'])
    .round(2)
    .sort_values('mean', ascending=False)
)

traffic_analysis
```

# **charts**

```python

import matplotlib.pyplot as plt

traffic_analysis['mean'].plot(kind='bar', figsize=(8, 5))

plt.title('Average Food Delivery Time by Traffic Density')
plt.xlabel('Road Traffic Density')
plt.ylabel('Average Delivery Time (minutes)')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

# **Correlation between numerical variables**

```python

numeric_columns = [
    'Delivery_person_Age',
    'Delivery_person_Ratings',
    'Vehicle_condition',
    'multiple_deliveries',
    'Time_taken (min)'
]

correlation = df_clean[numeric_columns].corr()

plt.figure(figsize=(9, 6))

sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    vmin=-1,
    vmax=1
)

plt.title('Correlation Heatmap of Delivery Variables')
plt.tight_layout()
plt.show()
```

# **Daily average delivery time**

```python

df_clean['Order_Date'] = pd.to_datetime(
    df_clean['Order_Date'],
    dayfirst=True,
    errors='coerce'
)

daily_delivery = (
    df_clean.dropna(subset=['Order_Date'])
    .groupby('Order_Date')['Time_taken (min)']
    .mean()
    .sort_index()
)

plt.figure(figsize=(12, 5))

plt.plot(
    daily_delivery.index,
    daily_delivery.values,
    marker='o'
)

plt.title('Daily Average Food Delivery Time')
plt.xlabel('Order Date')
plt.ylabel('Average Delivery Time (minutes)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

# **Final business insights and KPIs**

# **Calculate key metrics**

```python

# Key Performance Indicators (KPIs)

total_orders = df_clean['ID'].nunique()
average_delivery_time = df_clean['Time_taken (min)'].mean()
median_delivery_time = df_clean['Time_taken (min)'].median()
fastest_delivery = df_clean['Time_taken (min)'].min()
longest_delivery = df_clean['Time_taken (min)'].max()

print("Total Orders:", total_orders)
print("Average Delivery Time:", round(average_delivery_time, 2), "minutes")
print("Median Delivery Time:", median_delivery_time, "minutes")
print("Fastest Delivery:", fastest_delivery, "minutes")
print("Longest Delivery:", longest_delivery, "minutes")
```

# **Find the factors associated with longer delivery times**

```python

# Compare average delivery times across categories

factors = [
    'Road_traffic_density',
    'Weather_conditions',
    'City',
    'Type_of_vehicle',
    'Type_of_order',
    'Festival'
]

for col in factors:
    print("\nAverage delivery time by", col)
    print(
        df_clean.groupby(col)['Time_taken (min)']
        .mean()
        .round(2)
        .sort_values(ascending=False)
    )
```

# **Create a KPI chart**

```python

import matplotlib.pyplot as plt

kpi_names = [
    'Average',
    'Median',
    'Fastest',
    'Longest'
]

kpi_values = [
    average_delivery_time,
    median_delivery_time,
    fastest_delivery,
    longest_delivery
]

plt.figure(figsize=(8, 5))
plt.bar(kpi_names, kpi_values)

plt.title('Food Delivery Performance KPIs')
plt.xlabel('Metric')
plt.ylabel('Delivery Time (minutes)')

for i, value in enumerate(kpi_values):
    plt.text(i, value + 0.5, str(round(value, 2)), ha='center')

plt.tight_layout()
plt.show()
```

# **Generate an automatic business summary**

```python

traffic_avg = df_clean.groupby(
    'Road_traffic_density'
)['Time_taken (min)'].mean()

weather_avg = df_clean.groupby(
    'Weather_conditions'
)['Time_taken (min)'].mean()

city_avg = df_clean.groupby(
    'City'
)['Time_taken (min)'].mean()

print("FOOD DELIVERY ANALYSIS: BUSINESS SUMMARY")
print("-----------------------------------------")
print(f"Total unique orders: {total_orders}")
print(f"Average delivery time: {average_delivery_time:.2f} minutes")
print(f"Median delivery time: {median_delivery_time:.2f} minutes")
print(f"Highest average by traffic: {traffic_avg.idxmax()}")
print(f"Highest average by weather: {weather_avg.idxmax()}")
print(f"Highest average by city: {city_avg.idxmax()}")
```
